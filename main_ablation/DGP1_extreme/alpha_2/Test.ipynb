```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
file_path = "dataset_1_N250.csv"  # Update this path if needed
data = pd.read_csv(file_path)

# Scatter plot of column 'e' vs 'b'
plt.figure(figsize=(8, 6))
plt.scatter(data['b'], data['e'], alpha=0.5)
plt.xlabel(r'$b(X)$')
plt.ylabel(r'$\pi(X)$')
plt.grid(True)
plt.show()
```
